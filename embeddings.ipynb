```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
# Create sentences
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]

sent
```

```python
sent
```

```python
# Define vocabulary size
vocab_size = 10000
```

```python
one_hot_repr = [one_hot(words,vocab_size) for words in sent]
one_hot_repr

# one_hot() converts each word in a sentence into an integer ID
```

```python
sent
```

```python
# word Embedding Representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
sent_length = 8
embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_length)
print(embedded_docs)

# pad_sequences -> Makes all sentences the same length.
```

```python
# feature representation
dim = 10 # Represent every word using a vector of 10 numbers.
```

```python
model = Sequential()
# Add an Embedding layer:
model.add(Embedding(vocab_size,dim,input_length=sent_length))
# voc_size = total vocabulary size (10,000 possible word IDs)
# dim = size of the vector used to represent each word (10 numbers)
# input_length = number of word positions in each sentence (8)

model.compile('adam','mse')
```

```python
# Convert all padded sentences into 10-dimensional word embeddings.
model.predict(embedded_docs)
```
